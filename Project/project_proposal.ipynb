Magnetic confinement fusion devices are the leading candidate for the realization of commercial fusion energy, however, they are still not physically well understood and have many engineering challenges to overcome to reach commercialization. In magnetic confinement fusion devices, a common metric to evaluate various physical aspects of a device or configuration is taking a bounce average of specific quantities, such as the drift radially through the device, for particles that are bouncing in a magnetic well back and forth. These metrics can then be used for device improvements via optimization. The bouncing particles are called trapped particles as they are trapped in their bounce orbit. Many fusion energy codes, such as DESC (https://desc-docs.readthedocs.io/en/stable/), deploy quadrature strategies to evaluate the bounce average. In this project, myself and my partner would like to recreate some of these quadrature methods to evaluate bounce average in fusion devices for multiple magnetic well bounce points of particles at different locations in a device. A bounce averaged quantity has an integral with the form:

$$
\left \langle g \right \rangle = 2 \int_{t_i}^{t_f}g dt
$$

where the integral is evaluated between the initial ($t_i$) and final ($t_f$) points in the bounce orbit (over one half of the bounce, hence the factor of two, assuming the bounce is symmetric).

DESC is developed with the intention of optimization, thus it uses the jax python package to deploy auto-differentiation to make optimization plausible. With jax, normal scipy and numpy methods of quadratures are not usable and often must be written from scratch. This seems to be an ideal problem to tackle for a classroom setting, as we can deploy a quadrature method from scratch and apply it to a specific problem (the problem of bounce integration). However, due to the steep learning curve of jax, we will not use it. Instead, we will use numpy to develop a quadrature method and bounce-average quantities in a magnetic confinement fusion device. We will compare our bounce-integrated quantities to the outputs of DESC to verify the physics in this code.
