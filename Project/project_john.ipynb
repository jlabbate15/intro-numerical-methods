Magnetic confinement fusion devices are the leading candidate for the realization of commercial fusion energy, however, they are still not physically well understood and have many engineering challenges to overcome to reach commercialization. In magnetic confinement fusion devices, a common metric to evaluate various physical aspects of a device or configuration is taking a bounce average of specific quantities such as the drift radially through the device for particles that are trapped in a magnetic well. These metrics can then be used for device improvements via optimization. The bouncing particles are called trapped particles as they are trapped in the magnetic well.

In this project, we would like to recreate some of these quadrature methods to evaluate bounce averages in fusion devices for multiple magnetic-well bounce points of particles at different locations in a device. A bounce-averaged quantity has an integral of the form  
$$
\langle g \rangle = 2 \int_{t_i}^{t_f} g\, dt,
$$  
where the integral is evaluated between the initial $t_i$ and final $t_f$ points in the bounce orbit (over one half of the bounce, hence the factor of two, assuming the bounce is symmetric). In this Jupyter notebook we used methods from APMA4300 like cubic splines for the interpolation of the magnetic field in between known points and Gauss–Legendre quadrature methods to compute the bounce integrals. These techniques allow us to resolve the rapid spatial variation of the magnetic field along particle trajectories with high smoothness and numerical stability. By constructing accurate spline representations of the field, we ensure that the integrands used in the bounce calculations behave well under differentiation and integration. The use of Gauss–Legendre quadrature further improves the precision of the bounce-averaged quantities, enabling efficient evaluation even in regions where the particle motion becomes highly nonlinear.

DESC is a computational code designed to solve fixed-boundary ideal magnetohydrodynamic (MHD) equilibria in stellarators using a pseudo-spectral numerical method. The code employs a global Fourier–Zernike polynomial basis to represent the magnetic flux surfaces, ensuring analyticity at the magnetic axis and providing efficient convergence properties.

The flux surface coordinates $R(\rho, \theta, \zeta)$ and $Z(\rho, \theta, \zeta)$ are expanded in this basis as

$$
R(\rho, \theta, \zeta) = \sum_{n=-N}^{N} \sum_{m=-M}^{M} \sum_{l \in L} R_{lmn}\;\mathcal{Z}_{l}^{m}(\rho, \theta)\;\mathcal{F}^{n}(\zeta)
$$

$$
Z(\rho, \theta, \zeta) = \sum_{n=-N}^{N} \sum_{m=-M}^{M} \sum_{l \in L} Z_{lmn}\;\mathcal{Z}_{l}^{m}(\rho, \theta)\;\mathcal{F}^{n}(\zeta)
$$

where $\rho$ is the normalized radial coordinate, $\theta$ is the poloidal angle, and $\zeta$ is the toroidal angle. These coordinates are defined for a toroidal geometry. Here, $\mathcal{F}^{n}(\zeta) = e^{in\zeta}$ represents the toroidal Fourier modes, and $\mathcal{Z}_{l}^{m}(\rho, \theta) = R_{l}^{m}(\rho) e^{im\theta}$ denotes the Zernike polynomials defined on the unit disk.

The radial Zernike polynomials $R_{l}^{m}(\rho)$ are given by

$$
R_{l}^{m}(\rho) = \sum_{s=0}^{(l - |m|)/2} \frac{(-1)^s (l-s)!}{s! \left(\frac{l + |m|}{2} - s\right)! \left(\frac{l - |m|}{2} - s\right)!} \;\rho^{l-2s}
$$

where the index set

$$
L = \{ |m|,\, |m| + 2,\, |m| + 4,\, \ldots,\, L_{\text{max}} \}
$$

ensures that the radial dependence scales as $\rho^{|m|}$ near the magnetic axis, automatically satisfying the analyticity constraint and eliminating unphysical modes from the spectral representation.

We used cubic spline interpolation to ensure the interpolated magnetic fields were smooth and twice differentiable, which is important for accurately capturing curvature and avoiding sharp jumps in the first or second derivative. We want a well-behaved first and second derivative to ensure optimization is possible, as that is what these quantities are commonly used for. This method constructs piecewise cubic polynomials that join together with continuous first and second derivatives, producing a highly smooth approximation of the discretely sampled data. For numerical integration, we applied Gauss–Legendre quadrature, an efficient Gaussian quadrature technique that evaluates the integrand at optimally chosen points within each interval. Because these points and weights are selected to maximize accuracy, Gauss–Legendre quadrature provides very precise integral estimates even with a relatively small number of sampling points.

Below, we create an interpolation for the bounce integration. For an optimization loop - where this code would commonly be used - many optimizers require the second derivative of relavent quantities to be well behaved. In this class, we learned a few interpolation methods. This project will thus apply cubic splines for all interpolations.

```python
# Define a few helper functions that are jax differentiable
def polyfit_diff(x, y, deg): # jax.polyfit but differentiable 
    """
    Differentiable polynomial fit via least-squares.

    Args:
        x: shape (n,)
        y: shape (n,)
        deg: int, polynomial degree

    Returns:
        coeffs: shape (deg+1,), highest-power first
    """
    # Construct Vandermonde matrix (n, deg+1)
    A = jnp.vander(x, N=deg+1, increasing=False)

    # Solve least squares: minimize ||A c - y||^2
    coeffs, residuals, rank, s = jnp.linalg.lstsq(A, y, rcond=None)
    return coeffs

def affine_transform(x,a,b):
    """
    Map x from [a, b] to [-1, 1]

    Args:
        x: array
        a: lower bound of original interval
        b: upper bound of original interval
    """
    return 2 * (x - a) / (b - a) - 1

def arr_max_diff(arr):
    """
    Find the maximum of a 1D array (jax differentiable)

    Args:
        arr: 1D array

    Returns:
        dict with the index of max occurance and the value of the maximum
    """

    max_num = 99999999999999.9
    max_i = len(arr)+10 # this will throw an error if this function is not written correctly
    i_dumby = 0
    arr_dict = {'arr':arr,'i':i_dumby,'max_num':max_num,'max_i':max_i}
    def body_fun(i,arr_dict):
        cond_dict = {'arr':arr_dict['arr'],
                        'i':i,
                        'max_num':max_num,
                        'max_i':max_i}
        def tb_max(cond_dict):
            cond_dict['max_num'] = cond_dict['arr'][cond_dict['i']]
            cond_dict['max_i'] = cond_dict['i']
            return cond_dict
        def fb_max(cond_dict):
            return cond_dict
        return jax.lax.cond(cond_dict['arr'][i]<cond_dict['max_num'],tb_max,fb_max,cond_dict)
    out = jax.lax.fori_loop(0,len(arr),body_fun,arr_dict)
    return {'max_i':out['max_i'],'max_num':out['max_num']} # returns a dict with the index of max occurance and the value of the maximum
```

```python
# Interpolation using cubic splines
def interpolate(f_eval,x,method='cubic splines'):
    """
    Interpolate a function known only by the function evaluated at specific grid points

    Args:
        f_eval: array of f evaluated at points in x interval
        x: x points at which f was evaluated at to form f_eval

    Returns:
        d_arr_full: array of polynomial's second derivatives at each point
    """
        
    if method == 'cubic splines':
        x_affine = affine_transform(x,0,1)
        N = len(x_affine) # degree of polynomial is the number of data points we have - 1
        A = jnp.zeros((N-2,N-2))
        b = jnp.zeros(N-2)
        for i in range(0,N-2):
            k = i+1
            A = A.at[i,i].set(2)
            if i+1<N-2:
                A = A.at[i,i+1].set(8)
            if i+2<N-2:
                A = A.at[i,i+2].set(2)
            b = b.at[i].set(6*(f_eval[k+1] - f_eval[k-1]))

        # Solve for the second derivative coefficients
        d_arr = jnp.linalg.solve(A,b)
        # use "natural splines", specify second derivative at each boundary to be zero
        d_arr_full = jnp.zeros(N)
        d_arr_full = d_arr_full.at[0].set(0) # Boundary derivative
        d_arr_full = d_arr_full.at[1:N-1].set(d_arr)  # Interior derivatives
        d_arr_full = d_arr_full.at[-1].set(0) # Boundary derivative

        def evaluate_interpolated(x_new):
            """
            Evaluate a function at a new point using cubic splines

            Args:
                x_new: array of new points to evaluate the function at
                x: x points at which f was evaluated at to form f_eval
                f_eval: array of f evaluated at points in x interval

            Returns:
                P: array of values of the function at x_new
            """

            # Compute cubic spline coefficients
            d_star = d_arr_full
            
            results = jnp.zeros(len(x_new))
            for i in range(len(x_new)):

                # Find which interval contains x_new[i]
                k = arr_max_diff(-1*abs(x - x_new[i]))['max_i']
                
                # Get interval data
                # x_k = x[k]
                # x_kp1 = x[k + 1]
                y_k = f_eval[k] # y_k
                y_kp1 = f_eval[k + 1] # y_{k+1}
                d_k = d_star[k] # d^*_k
                d_kp1 = d_star[k + 1] # d^*_{k+1}

                s = affine_transform(x_new[i],0,1) # this is just one value
                
                # Evaluate cubic polynomial using the formula from lecture 06
                P = ((1-s)**2 * (1 + 2*s) * y_k + s**2 * (3 - 2*s) * y_kp1 + s * (1-s)**2 * d_k - s**2 * (1-s) * d_kp1)
                results = results.at[i].set(P)
            return results
        return evaluate_interpolated

    else:
        raise ValueError("Invalid interpolation method") 
```

Since particles are assumed to be stuck to the magnetic field lines, the shape of the magnitude of the magnetic field curve along a field line will determine any mirroring or trapping that occurs for a particle. The magnitude of the magnetic field $B_c$ that causes a particle to turn around (an effect called mirroring [3]) is set by a parameter known as the pitch parameter $\lambda$.
$$
B_c = \frac{1}{\lambda}
$$
A routine is written in the following code cell to specify a grid of $B_c$ depending on the maximum and minimum values of the magnetic field magnitude for one flux surface ($\rho$), magnetic field line ($\alpha$), and magnetic well combination. This routine ensures that each particle is "trapped" in that it mirrors back and forth within a magnetic well (or "barely-trapped" if there is only one mirror point).

```python
# Specify grid for B_c
from desc.equilibrium.coords import get_rtz_grid

def get_Bc_grid(eq,rho,num_Bc):
    """
    Get B_crit values for each rho considered

    Args:
        eq: DESC equilibrium
        rho: array of rho values
        num_Bc: integer number of B_crit values to consider

    Returns:
        2D array of B_crit values to consider, shape: (rho,Bc)
    """
    rho_bc = rho
    alpha_bc = jnp.array([0])
    zeta_bc = jnp.linspace(0, 12 * jnp.pi, 50)
    grid_bc = get_rtz_grid(eq, rho_bc, alpha_bc, zeta_bc, coordinates="raz")
    data = eq.compute(["min_tz |B|", "max_tz |B|"], grid=grid_bc) # min,max |B| on each rho
    return jnp.linspace(
        jnp.mean(data['min_tz |B|']),
        jnp.mean(data['max_tz |B|']),
        num_Bc
    )
```

The next cell determines the spatial point at which particle mirroring occurs is recorded for each combination of $\rho,\alpha,B_c,$ and magnetic well along the relevant field line. For simplicity, we will use the DESC method to do this. A method could be written using a root-finder, however, it will likely make this project too long.

```python
# Identifying bounce/mirror points
from desc.integrals.bounce_integral import Bounce1D
from desc.integrals.quad_utils import (
    automorphism_sin,
    get_quadrature,
    grad_automorphism_sin,
)
from orthax.legendre import leggauss
from desc.grid import LinearGrid
from desc.grid import Grid

def get_bounce_points(eq,grid,Bcs,num_quad=32,num_well=3):
    """
    Find zeta bounce points for each rho,alpha,well,Bc considered

    Args:
        eq: DESC equilibrium
        grid: relevant grid
        Bcs: array of B_crits to consider
        num_quad: integer number of quadrature points to consider
        num_well: integer number of magnetic wells to consider

    Returns:
        array of zeta bounce point values to consider for each rho,alpha,well,Bc. shape (rho,alpha,well,Bc,zeta one and two)
    """

    # DESC method
    rho = jnp.unique(grid.nodes[:,0])
    _keys_1dr = ["iota", "iota_r", "min_tz |B|", "max_tz |B|"]
    _grid_1dr = LinearGrid(
            rho=rho, M=eq.M_grid, N=eq.N_grid, NFP=eq.NFP, sym=eq.sym
    )
    rho_vals = rho
    alpha_vals = jnp.linspace(0,2*jnp.pi,num_alpha)
    zeta_vals = jnp.linspace(0,10,15)
    bounce_grid = Grid.create_meshgrid(
        [rho_vals, alpha_vals, zeta_vals],
        coordinates="raz",
        NFP=eq.NFP
    )
    data = {
         'B^zeta': eq.compute("B^zeta",grid=bounce_grid)['B^zeta'],
         'B^zeta_z|r,a': eq.compute("B^zeta_z|r,a",grid=bounce_grid)['B^zeta_z|r,a'], 
         '|B|': eq.compute("|B|",grid=bounce_grid)['|B|'], 
         '|B|_z|r,a': eq.compute("|B|_z|r,a",grid=bounce_grid)['|B|_z|r,a']
    }
    quad = (
        get_quadrature(
            leggauss(num_quad),
            (automorphism_sin, grad_automorphism_sin),
        )
    )
    bounce = Bounce1D(bounce_grid, data, quad, is_reshaped=False)
    return bounce.points(Bcs, num_well=num_well)
```

Once each bounce point/mirror point has been identified, the actual integration over a period of mirroring may be calculated.
$$
\langle f \rangle_b = \frac{2}{v\tau_b}\int_{l_0}^{l_f}\frac{f}{\sqrt{1-\frac{B}{B_c}}}dl
$$
This integral will be calculated for each $\rho,\alpha,B_c,$ and magnetic well combination using the Gauss-Legendre quadrature method. The quadrature weights and associated quantities are defined in the following cell.

```python
# Create quadratures
def get_quad_leggaus(rho,Bcs,points,num_quad=32):
    """
    Find quadrature points using the Gauss-Legendre quadrature

    Args:
        points: array of bounce points
        num_quad: number of quadrature points

    Returns:
        B, B_sup_zeta evaluated at quadrature points
    """
    z1 = points[0][:][:][:][:] # rho, alpha, Bc, well
    z2 = points[1][:][:][:][:] # rho, alpha, Bc, well
    x_ref, w_ref = leggauss(num_quad)
    x_ref = jnp.broadcast_to(x_ref[None,None,None,None,...],(z1.shape[0],z1.shape[1],z1.shape[2],z1.shape[3],num_quad))
    z1 = jnp.broadcast_to(z1[...,None],(z1.shape[0],z1.shape[1],z1.shape[2],z1.shape[3],num_quad))
    z2 = jnp.broadcast_to(z2[...,None],(z2.shape[0],z2.shape[1],z2.shape[2],z2.shape[3],num_quad))
    zeta = 0.5*(z1 + z2) + 0.5*(z2 - z1) * x_ref # shape (rho, alpha, Bc, well, num_quad)
    dzetadx  = 0.5 * (z2 - z1) # shape (rho, alpha, Bc, well, num_quad)
    
    w_ref = jnp.broadcast_to(w_ref[None,None,None,None,...],(z1.shape[0],z1.shape[1],z1.shape[2],z1.shape[3],num_quad))
    return zeta, w_ref, dzetadx  # integrate: sum(w_ref * f(zeta) * x_to_zeta_jac)
```

To now develop the integral numerically, we must keep all our variable changes in order. We can write this integral shown above with $g = \frac{f}{\sqrt{1-\frac{B}{B_c}}}$ in a way that allows the use of the quantities we have considered in this document so far:

$$
\int_{l_0}^{l_f}g(l)dl = \int_{\zeta_1}^{\zeta_2}g(\zeta)\frac{dl}{d\zeta}d\zeta = \int_{x_1}^{x_2}g(x)\frac{dl}{d\zeta}\frac{d\zeta}{dx}dx 
$$

where $x=[-1,1]$ defines the grid of quadratures.

$g$ is some callable function that will be defined later in this code. 

$\frac{d\zeta}{dx}$ is the Jacobian from $\zeta$ to $x$ which has been defined in the get_quad_leggaus() method above.

$\frac{dl}{d\zeta}$ is known from [2] to be:

$$
\frac{dl}{d\zeta} = \left| \frac{\mathbf{B}}{\mathbf{B} \cdot \mathbf{\nabla} \zeta} \right| = \frac{|\mathbf{B}|}{|\mathbf{B^\zeta}|} = \frac{1}{|\mathbf{b^\zeta}|}
$$

These parameters may be calculated at the quadrature points using the following code.

```python
from jax import vmap
def compute_Bs_quad(eq,grid,zeta_quad):
    """
    Find B and B_sup_zeta at quadrature zeta points

    Args:
        eq: DESC equilibrium
        grid: relevant grid
        zeta_quad: zeta quadrature points

    Returns:
        B, B_sup_zeta evaluated at quadrature points
    """
    rhos_shape = len(jnp.unique(grid.nodes[:,0]))
    alpha_shape = num_alpha
    zeta_shape = len(jnp.unique(grid.nodes[:,2]))
    # B, B_sup_zeta = compute_from_eq(eq=eq,grid=grid) # 1D arrays but should be functions of (rho,alpha,zeta)
    # B = jnp.broadcast_to(B['|B|'][None,None,None],(rhos_shape,alpha_shape,zeta_shape))
    # B_sup_zeta = jnp.broadcast_to(B_sup_zeta['B^zeta'][None,None,None],(rhos_shape,alpha_shape,zeta_shape))
    # zeta_knots = jnp.unique(grid.nodes[:,2]) # zeta coordinates, 1D array

    # # Find B at quadrature points
    # B_spline = interpolate(B, zeta_knots, method='cubic splines') # Build B spline
    # B_quad = B_spline(x_new=zeta_quad) # B at quadrature points for each spatial location

    # # Find B_sup_zeta at quadrature points
    # B_sup_zeta_spline = interpolate(B_sup_zeta, zeta_knots, method='cubic splines') # Build B_sup_zeta spline
    # B_sup_zeta_quad = B_sup_zeta_spline(x_new=zeta_quad) # 1D array of B_sup_zeta at quadrature points

    # return B_quad, B_sup_zeta_quad

    # Extract unique zeta coordinates
    zeta_knots = jnp.unique(grid.nodes[:, 2])  # zeta coordinates, 1D array
    
    # Get B and B_sup_zeta data from equilibrium
    B_data, B_sup_zeta_data = compute_from_eq(eq=eq, grid=grid)
    
    # Extract arrays from dictionaries and reshape to (num_rho, num_alpha, num_zeta)
    B_3d = B_data['|B|'].reshape((zeta_shape, rhos_shape, alpha_shape)).transpose(1,2,0)
    B_sup_zeta_3d = B_sup_zeta_data['B^zeta'].reshape((zeta_shape, rhos_shape, alpha_shape)).transpose(1,2,0)

    # Define function to interpolate for a single (rho, alpha) field line
    def interpolate_single_fieldline(B_1d, zeta_quad_1d):
        """Interpolate B along a single field line at given quadrature points"""
        B_spline = interpolate(B_1d, zeta_knots, method='cubic splines')
        return B_spline(x_new=zeta_quad_1d)
    
    # Define function to handle (Bc, well) dimensions for one (rho, alpha)
    def interpolate_for_bc_well(B_1d, zeta_quad_bc_well):
        """
        Interpolate for one (rho, alpha) at all (Bc, well) quadrature points
        
        Args:
            B_1d: B values for one (rho, alpha), shape (num_zeta,)
            zeta_quad_bc_well: quadrature points, shape (num_Bc, num_well, num_quad)
        
        Returns:
            B interpolated, shape (num_Bc, num_well, num_quad)
        """
        # vmap over (Bc, well) dimensions
        return vmap(
            vmap(interpolate_single_fieldline, in_axes=(None, 0)), 
            in_axes=(None, 0))(B_1d, zeta_quad_bc_well)
    
    # vmap over (rho, alpha) axes
    # B_3d: (num_rho, num_alpha, num_zeta), zeta_quad: (num_rho, num_alpha, num_Bc, num_well, num_quad)
    interpolate_over_ra = vmap(
        vmap(interpolate_for_bc_well, in_axes=(0, 0)), 
        in_axes=(0, 0))
    
    # Apply interpolation
    B_quad = interpolate_over_ra(B_3d, zeta_quad)  # Shape: (num_rho, num_alpha, num_Bc, num_well, num_quad)
    B_sup_zeta_quad = interpolate_over_ra(B_sup_zeta_3d, zeta_quad)  # zeta_quad shape: (rho,alpha,Bc,well,num_quad), Output shape: (num_rho, num_alpha, num_Bc, num_well, num_quad)
    
    return B_quad, B_sup_zeta_quad
```

```python
# Integrate!
def bounce_integrate(eq,grid,f,num_well,num_Bc,num_quad):
    """
    Integrate f over a bounce period

    Args:
        eq: DESC equilibrium
        grid: relevant user-specifiedgrid
        f: callable function to be integrated
        num_well: integer number of magnetic wells to consider
        num_Bc: integer number of Bc to consider
        num_quad: integer number of quadrature points to consider

    Returns:
        array of bounce integrated values of f size (rho,alpha,Bc,well)
    """

    rho = jnp.unique(grid.nodes[:, 0])
    Bcs = get_Bc_grid(eq,rho,num_Bc=num_Bc) # determine critical magnetic field values
    bounce_points = get_bounce_points(eq,grid,Bcs,num_well=num_well) # determine bounce points based on B_crits
    zeta_quad, w_ref, dzetadx = get_quad_leggaus(rho,Bcs,bounce_points,num_quad=num_quad) # determine zeta quadrature points, output shapes are (rho,alpha,Bc,well,num_quad)
    B_quad, B_sup_zeta_quad = compute_Bs_quad(eq,grid,zeta_quad) # determine B and B_sup_zeta at quadrature points
    dldzeta = B_quad/B_sup_zeta_quad
    data_null = {'null':0} # dummy, not used, but needed input for pure DESC case
    Bcs = jnp.broadcast_to(Bcs[None,None,...,None,None],(B_quad.shape[0],B_quad.shape[1],B_quad.shape[2],B_quad.shape[3],B_quad.shape[4]))

    return jnp.sum(f(data_null,B_quad,pitch=Bcs**-1) * dldzeta * dzetadx * w_ref,axis=-1) # should have an output shape (rho,alpha,Bc,well)
```
